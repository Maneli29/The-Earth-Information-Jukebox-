```python
import os
import cv2

FILES = [
    r"C:\Users\Quantam Waffle\OneDrive\Documents\GitHub\The-Earth-Information-Jukebox-\data\elnino_avw_2025_2026_daily_1900x800.mp4",
    r"C:\Users\Quantam Waffle\OneDrive\Documents\GitHub\The-Earth-Information-Jukebox-\data\iss_lightning_sphere_07312023.mp4",
]  

OUT = "data/samples"
os.makedirs(OUT, exist_ok=True)
IMAGE_EXT = (".png", ".jpg", ".jpeg", ".tif", ".tiff")

for path in FILES:
    name = os.path.splitext(os.path.basename(path))[0]
    if not os.path.exists(path):
        print("NOT FOUND:", path)
        continue

    if path.lower().endswith(IMAGE_EXT):
        img = cv2.imread(path)
        print(path, "-> image,", img.shape[1], "x", img.shape[0])
        cv2.imwrite(f"{OUT}/{name}_copy.png", img)
        continue

    cap = cv2.VideoCapture(path)
    w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    n = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    print(path, "->", w, "x", h, "|", round(fps, 1), "fps |",
          n, "frames |", round(n / fps, 1) if fps else "?", "seconds")

    for k in range(4):
        idx = int(n * k / 4)
        cap.set(cv2.CAP_PROP_POS_FRAMES, idx)
        ok, frame = cap.read()
        if ok:
            cv2.imwrite(f"{OUT}/{name}_frame{idx}.png", frame)
    cap.release()

print("Done. Look in data/samples")
```
